# 03 — Preparación del dataset para modelado

Este notebook parte de `DataSet_depurado_basico_EDA.xlsx` y genera `DataSet_modelado_base.xlsx`, que será la entrada de los notebooks de entrenamiento.

Las exclusiones se realizan de forma explícita y documentada:

- Columnas 100% vacías;
- Identificadores;
- Domicilio de granularidad excesiva;
- Texto libre no estructurado;
- Variables sin variabilidad.

La variable nota se conserva para construir posteriormente las vistas de clasificación.

## 1. Importación de librerías

```python
import pandas as pd
import numpy as np

from google.colab import files
from IPython.display import display

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 200)

print("Librerías cargadas correctamente.")
```

## 2. Carga del dataset depurado básico obtenido en el notebook 02

```python
uploaded = files.upload()

if not uploaded:
    raise ValueError("No se seleccionó ningún archivo.")

ruta_archivo = next(iter(uploaded))
df_depurado = pd.read_excel(ruta_archivo)

print(f"Archivo cargado: {ruta_archivo}")
print(f"Filas: {df_depurado.shape[0]}")
print(f"Columnas: {df_depurado.shape[1]}")

display(df_depurado.head())
```

## 3. Configuración de exclusiones

```python
variable_objetivo_original = "nota"

columnas_identificatorias = [
    "alumno"
]

columnas_domicilio_granular = [
    "procedencia_calle",
    "procedencia_numero",
    "procedencia_piso",
    "procedencia_departamento",
    "procedencia_unidad",
    "procedencia_barrio"
]

columnas_texto_libre = [
    "trabajo_tarea_descripcion",
    "costeo_estudios_descripcion",
    "idioma_otro",
    "nivel_idioma_otro",
    "otra_descripcion",
    "informacion_importante"
]

variables_territoriales_a_revisar = [
    "procedencia_localidad",
    "procedencia_codigo_postal"
]

if variable_objetivo_original not in df_depurado.columns:
    raise ValueError(
        f"No se encontró la columna '{variable_objetivo_original}'."
    )

print("Configuración definida correctamente.")
```

## 4. Normalización y validación de la variable objetivo

```python
df_depurado[variable_objetivo_original] = (
    df_depurado[variable_objetivo_original]
    .astype("string")
    .str.strip()
    .str.upper()
)

valores_esperados = {"A", "R", "U"}
valores_observados = set(
    df_depurado[variable_objetivo_original]
    .dropna()
    .unique()
)

valores_no_esperados = valores_observados - valores_esperados

if valores_no_esperados:
    print(
        "Advertencia: valores de nota no esperados:",
        valores_no_esperados
    )
else:
    print("Correcto: nota contiene únicamente A, R y U.")

distribucion_nota = (
    df_depurado[variable_objetivo_original]
    .value_counts(dropna=False)
    .rename_axis(variable_objetivo_original)
    .reset_index(name="cantidad")
)

distribucion_nota["porcentaje"] = (
    distribucion_nota["cantidad"] / len(df_depurado) * 100
).round(2)

display(distribucion_nota)
```

## 5. Inicialización del dataset de modelado

```python
df_modelado_base = df_depurado.copy()

dimensiones_iniciales = df_modelado_base.shape
print("Dimensiones iniciales:", dimensiones_iniciales)
```

## 6. Eliminación de columnas 100% vacías

```python
columnas_100_faltantes = df_modelado_base.columns[
    df_modelado_base.isna().all()
].tolist()

df_modelado_base = df_modelado_base.drop(
    columns=columnas_100_faltantes
)

print("Columnas eliminadas:")
print(columnas_100_faltantes)
print("Dimensiones actuales:", df_modelado_base.shape)
```

## 7. Exclusión de identificadores

```python
columnas_identificatorias_existentes = [
    col for col in columnas_identificatorias
    if col in df_modelado_base.columns
]

df_modelado_base = df_modelado_base.drop(
    columns=columnas_identificatorias_existentes
)

print("Identificadores eliminados:")
print(columnas_identificatorias_existentes)
print("Dimensiones actuales:", df_modelado_base.shape)
```

## 8. Exclusión de variables de domicilio granular

```python
columnas_domicilio_existentes = [
    col for col in columnas_domicilio_granular
    if col in df_modelado_base.columns
]

df_modelado_base = df_modelado_base.drop(
    columns=columnas_domicilio_existentes
)

print("Variables de domicilio granular eliminadas:")
print(columnas_domicilio_existentes)
print("Dimensiones actuales:", df_modelado_base.shape)
```

## 9. Exclusión de texto libre no estructurado

```python
columnas_texto_existentes = [
    col for col in columnas_texto_libre
    if col in df_modelado_base.columns
]

df_modelado_base = df_modelado_base.drop(
    columns=columnas_texto_existentes
)

print("Variables de texto libre eliminadas:")
print(columnas_texto_existentes)
print("Dimensiones actuales:", df_modelado_base.shape)
```

## 10. Exclusión de variables sin variabilidad

```python
columnas_sin_variabilidad = []

for col in df_modelado_base.columns:
    if col == variable_objetivo_original:
        continue

    if df_modelado_base[col].dropna().nunique() <= 1:
        columnas_sin_variabilidad.append(col)

df_modelado_base = df_modelado_base.drop(
    columns=columnas_sin_variabilidad
)

print("Variables sin variabilidad eliminadas:")
print(columnas_sin_variabilidad)
print("Dimensiones actuales:", df_modelado_base.shape)
```

## 11. Revisión de variables territoriales conservadas

Estas variables no se eliminan automáticamente. Se muestran para evaluar su cardinalidad y decidir si deben mantenerse, agruparse o excluirse posteriormente.

```python
variables_territoriales_existentes = [
    col for col in variables_territoriales_a_revisar
    if col in df_modelado_base.columns
]

for col in variables_territoriales_existentes:
    print("\nVariable:", col)
    print(
        "Valores únicos:",
        df_modelado_base[col].nunique(dropna=True)
    )
    display(
        df_modelado_base[col]
        .value_counts(dropna=False)
        .head(20)
    )
```

## 12. Control de variables de alta cardinalidad

```python
registros_cardinalidad = []

for col in df_modelado_base.columns:
    if col == variable_objetivo_original:
        continue

    registros_cardinalidad.append({
        "variable": col,
        "tipo_dato_pandas": str(df_modelado_base[col].dtype),
        "valores_unicos": df_modelado_base[col].nunique(dropna=True),
        "porcentaje_faltantes": round(
            df_modelado_base[col].isna().mean() * 100,
            2
        )
    })

resumen_cardinalidad = (
    pd.DataFrame(registros_cardinalidad)
    .sort_values("valores_unicos", ascending=False)
    .reset_index(drop=True)
)

print("Variables con mayor cardinalidad:")
display(resumen_cardinalidad.head(30))
```

## 13. Resumen del dataset final

```python
filas, columnas = df_modelado_base.shape
total_celdas = filas * columnas
total_faltantes = int(df_modelado_base.isna().sum().sum())

porcentaje_faltantes = (
    total_faltantes / total_celdas * 100
    if total_celdas else 0
)

resumen_modelado_base = pd.DataFrame({
    "concepto": [
        "Filas",
        "Columnas finales",
        "Celdas totales",
        "Valores faltantes",
        "Porcentaje de valores faltantes",
        "Columnas 100% vacías eliminadas",
        "Identificadores eliminados",
        "Domicilio granular eliminado",
        "Texto libre eliminado",
        "Variables sin variabilidad eliminadas"
    ],
    "valor": [
        filas,
        columnas,
        total_celdas,
        total_faltantes,
        round(porcentaje_faltantes, 2),
        len(columnas_100_faltantes),
        len(columnas_identificatorias_existentes),
        len(columnas_domicilio_existentes),
        len(columnas_texto_existentes),
        len(columnas_sin_variabilidad)
    ]
})

display(resumen_modelado_base)

print("\nColumnas finales:")
for col in df_modelado_base.columns:
    print("-", col)
```

## 14. Verificación final de la variable objetivo

```python
if variable_objetivo_original not in df_modelado_base.columns:
    raise ValueError(
        "La variable objetivo fue eliminada por error."
    )

print(
    f"Correcto: '{variable_objetivo_original}' se conserva "
    "para construir las vistas de modelado."
)

display(
    df_modelado_base[variable_objetivo_original]
    .value_counts(dropna=False)
    .rename_axis(variable_objetivo_original)
    .reset_index(name="cantidad")
)
```

## 15. Exportación del dataset de modelado base

```python
nombre_salida_excel = "DataSet_modelado_base.xlsx"

with pd.ExcelWriter(nombre_salida_excel, engine="openpyxl") as writer:
    df_modelado_base.to_excel(
        writer,
        sheet_name="Datos_modelado_base",
        index=False
    )

    resumen_modelado_base.to_excel(
        writer,
        sheet_name="Resumen_preparacion",
        index=False
    )

    resumen_cardinalidad.to_excel(
        writer,
        sheet_name="Cardinalidad",
        index=False
    )

print(f"Archivo generado: {nombre_salida_excel}")
files.download(nombre_salida_excel)
```
